# TAREA 12: DISTRIBUCIONES A PRIORI Y POSTERIORI

a) Calcule las probabilidades a priori (P(Ci)) para cada clase.

```python
# Datos de la tabla: lista de etiquetas "Robado" para cada registro
robado = ['Sí', 'No', 'Sí', 'No', 'Sí', 'No', 'Sí', 'No', 'No', 'Sí']

# Contar frecuencia de cada clase
from collections import Counter
frecuencias = Counter(robado)

# Total de registros
total = len(robado)

# Calcular probabilidades a priori
probabilidades = {clase: freq / total for clase, freq in frecuencias.items()}

# Mostrar resultados
for clase, prob in probabilidades.items():
    print(f"P({clase}) = {prob:.2f}")
```

b) Calcule las probabilidades condicionales necesarias.

```python
# Datos de la tabla: lista de diccionarios con atributos y clase (Robado)
data = [
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "Sí"},
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "No"},
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "Sí"},
    {"Color": "Amarillo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "No"},
    {"Color": "Amarillo", "Tipo": "Deportivo", "Origen": "Importado", "Robado": "Sí"},
    {"Color": "Amarillo", "Tipo": "SUV", "Origen": "Importado", "Robado": "No"},
    {"Color": "Amarillo", "Tipo": "SUV", "Origen": "Importado", "Robado": "Sí"},
    {"Color": "Amarillo", "Tipo": "SUV", "Origen": "Doméstico", "Robado": "No"},
    {"Color": "Rojo", "Tipo": "SUV", "Origen": "Importado", "Robado": "No"},
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Importado", "Robado": "Sí"},
]

# Registro a clasificar
X = {"Color": "Rojo", "Tipo": "SUV", "Origen": "Doméstico"}

# Contar total registros por clase
total_por_clase = {}
for registro in data:
    clase = registro["Robado"]
    total_por_clase[clase] = total_por_clase.get(clase, 0) + 1

# Función para contar frecuencia de atributo dado clase
def freq_atributo_dado_clase(atributo, valor, clase):
    return sum(1 for r in data if r["Robado"] == clase and r[atributo] == valor)

# Calcular probabilidades condicionales para cada clase
for clase in total_por_clase:
    print(f"Clase: {clase}")
    total = total_por_clase[clase]
    for atributo, valor in X.items():
        freq = freq_atributo_dado_clase(atributo, valor, clase)
        prob = freq / total
        print(f"P({atributo}={valor} | Robado={clase}) = {freq} / {total} = {prob:.2f}")
    print()
```

c) Calcule las probabilidades a posteriori (P(X|Ci)) y utilícelas para clasificar el registro X.

```python
# Datos de la tabla: lista de diccionarios con atributos y clase (Robado)
data = [
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "Sí"},
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "No"},
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "Sí"},
    {"Color": "Amarillo", "Tipo": "Deportivo", "Origen": "Doméstico", "Robado": "No"},
    {"Color": "Amarillo", "Tipo": "Deportivo", "Origen": "Importado", "Robado": "Sí"},
    {"Color": "Amarillo", "Tipo": "SUV", "Origen": "Importado", "Robado": "No"},
    {"Color": "Amarillo", "Tipo": "SUV", "Origen": "Importado", "Robado": "Sí"},
    {"Color": "Amarillo", "Tipo": "SUV", "Origen": "Doméstico", "Robado": "No"},
    {"Color": "Rojo", "Tipo": "SUV", "Origen": "Importado", "Robado": "No"},
    {"Color": "Rojo", "Tipo": "Deportivo", "Origen": "Importado", "Robado": "Sí"},
]

# Registro a clasificar
X = {"Color": "Rojo", "Tipo": "SUV", "Origen": "Doméstico"}

# Contar total registros por clase
total_por_clase = {}
for registro in data:
    clase = registro["Robado"]
    total_por_clase[clase] = total_por_clase.get(clase, 0) + 1

# Función para contar frecuencia de atributo dado clase
def freq_atributo_dado_clase(atributo, valor, clase):
    return sum(1 for r in data if r["Robado"] == clase and r[atributo] == valor)

# Calcular probabilidades a priori
prob_a_priori = {clase: total / len(data) for clase, total in total_por_clase.items()}

# Calcular probabilidades condicionales y a posteriori
posteriori = {}
for clase in total_por_clase:
    prob_condicional = 1
    for atributo, valor in X.items():
        freq = freq_atributo_dado_clase(atributo, valor, clase)
        prob = freq / total_por_clase[clase]
        prob_condicional *= prob
    posteriori[clase] = prob_a_priori[clase] * prob_condicional

# Mostrar resultados
for clase, prob in posteriori.items():
    print(f"P({clase} | X) ∝ {prob:.5f}")

# Clasificación
clase_predicha = max(posteriori, key=posteriori.get)
print(f"\nEl registro X se clasifica como: {clase_predicha}")
```
